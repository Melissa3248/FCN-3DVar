```python
# compute CRPS

def convert_array_of_idxs(array):
    return np.stack([convert_idx_to_latlon(array[i,0], array[i,1]) for i in range(array.shape[0])])
```

```python
from matplotlib.pyplot import figure


color = ['red','green', 'blue', 'cyan']
markers = ["v", 'o', 's', 'D']

def plot_extreme_event(res, gt, crps, res_color, res_label, figure_name):
    factor = 1
    figure(figsize=(8*factor, 6*factor), dpi=100)

    plt.imshow(d<0, cmap = "Greys")

    for n in range(50):
        for t in range(1,30):
            if t == 1 and n == 1:

                plt.plot([res[n,t-1,0], res[n,t,0]], [res[n,t-1,1],res[n,t,1]], c = res_color,alpha = 0.2, label = res_label)
            else:
                plt.plot([res[n,t-1,0], res[n,t,0]], [res[n,t-1,1],res[n,t,1]], c = res_color,alpha = 0.2)


    plt.legend()
    plt.xticks([])
    plt.yticks([])


    num_model = torch.load("/pscratch/sd/m/maadrian/data/ERA5_2023/2023/ECMWF_forecasts/mawar_forecast_ecmwf.pt")[1:]
    
    for t in range(30):
        

        # numerical weather model forecast from ECMWF
        
        if t !=0:
            plt.plot([num_model[t-1,0], num_model[t,0]], [num_model[t-1,1],num_model[t,1]], c = "magenta", linewidth = 0.5, zorder = 1000)
            plt.scatter(num_model[t,0], num_model[t,1], c = "magenta",  s = 12, marker='D', zorder = 100)
        else:
            plt.scatter(num_model[t,0], num_model[t,1], c = "magenta",  s = 12, marker='D', label = "IFS-HRES Forecast", zorder = 200)
        
        # ERA5 reanalysis
        if t !=0:
            plt.plot([gt[t-1,0], gt[t,0]], [gt[t-1,1],gt[t,1]], c = "cyan", linewidth = 0.5, zorder = 1000)
            plt.scatter(gt[t,0], gt[t,1], c = "cyan",  s = 12, marker=tcmarkers.HU, zorder = 100)
        else:
            plt.scatter(gt[t,0], gt[t,1], c = "cyan",  s = 12, marker=tcmarkers.HU, label = "ERA5 Reanalysis", zorder = 200)
    
        # IBTrACS data 
        if t !=0:
            plt.plot([real_data[t-1,0], real_data[t,0]], [real_data[t-1,1],real_data[t,1]], c = "gold", linewidth = 0.5, zorder = 1000)
            plt.scatter(real_data[t,0], real_data[t,1], c = "gold",  s = 12, marker='X', zorder = 100)
        else:
            plt.scatter(real_data[t,0], real_data[t,1], c = "gold",  s = 12, marker='X', label = "IBTrACS Data", zorder = 200)
        
    plt.legend( bbox_to_anchor=(1.02, 0))
    plt.title(f"Eye of Typhoon Mawar\n May 23, 2023 00:00 UTC - May 30, 2023 12:00 UTC\nCRPS: {np.round(crps,2)}$^\circ$", fontsize = 15)


gt = true_min_pressure_over_t
plot_extreme_event(analysis_res[:,1:,:], gt, crps_analysis, color[2], 
                   res_label = "3DVar Analysis ({}$^\circ$ Obs.) ".format(degrees) +"+ $\{N(0,0.3I_{d_x})\}^{1:50}$ Initialization", figure_name = "task3/extreme_typhoon_eye_{}deg_analysis".format(degrees))
```
