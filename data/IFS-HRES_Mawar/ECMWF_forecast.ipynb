```python
import torch
import tcmarkers
import h5py
import matplotlib.pyplot as plt


offset_lat = 100
offset_lon = 0#100
half_window = 50
m_u10=np.load('/pscratch/sd/m/maadrian/stats_v0/ERA5/global_means.npy')[0,0,0,0]
s_u10=np.load('/pscratch/sd/m/maadrian/stats_v0/ERA5/global_stds.npy')[0,0,0,0]

m_v10=np.load('/pscratch/sd/m/maadrian/stats_v0/ERA5/global_means.npy')[0,1,0,0]
s_v10=np.load('/pscratch/sd/m/maadrian/stats_v0/ERA5/global_stds.npy')[0,1,0,0]
print(m_v10, s_v10)
print((17 - m_v10)/s_v10)
n=2

forecast_min_pressure_over_t = []

forecast_max_ws = []

forecast_min_pressure = []

for i in range(41):
    print(i)
    if i <=5:
        offset_lat = 75
    else:
        offset_lat = 0

    truth_pressure = data[0,i,200:350,450+15:600+15]
    
    truth_v10 = data[1,i,200:350,450+15:600+15]

    truth_v10 = torch.from_numpy(truth_v10)
    
    truth_u10 = data[2,i,200:350,450+15:600+15]
    truth_u10 = torch.from_numpy(truth_u10)
    
    truth_ws = torch.sqrt(truth_v10*truth_v10 + truth_u10*truth_u10)
    
    truth_pressure = torch.from_numpy(truth_pressure)

    min_pressure = (truth_pressure[offset_lat:,:]==torch.min(truth_pressure[offset_lat:,:])).nonzero()
    if len(min_pressure) >1 :
        min_pressure = min_pressure[0]
    
    
    forecast_min_pressure_over_t.append([min_pressure.squeeze()[1], min_pressure.squeeze()[0]+offset_lat])

    w_max = torch.max(truth_ws)
    
    forecast_max_ws.append(w_max)
    forecast_min_pressure.append(torch.min(truth_pressure[offset_lat:,:]))

    plt.imshow(truth_pressure)
    plt.scatter([min_pressure.squeeze()[1]], [min_pressure.squeeze()[0]+offset_lat], c = 'r', marker=tcmarkers.HU)
    print([min_pressure.squeeze()[1]], [min_pressure.squeeze()[0]+offset_lat])
    print(w_max)

    print(w_max, torch.min(truth_ws))

    plt.show()
print(forecast_min_pressure_over_t)
forecast_min_pressure_over_t=torch.tensor(forecast_min_pressure_over_t)

```
